# Phase 3 - Step 15: Recommendation Engine

This notebook implements a transparent rule-based upskilling engine that maps missing skills to courses, certifications, hands-on projects, and prioritized learning paths.

```python
import pandas as pd
from pathlib import Path

processed_dir = Path("data/processed")
df_gaps = pd.read_csv(processed_dir / "employee_skill_gaps.csv")
df_org_gaps = pd.read_csv(processed_dir / "organization_skill_gaps.csv")

severity_map = df_org_gaps.set_index('missing_skill')['severity'].to_dict()

# Course, certification, and project catalog for upskilling
upskilling_catalog = {
    'Python': {
        'course': 'Advanced Python for Enterprise Systems',
        'certification': 'PCEP / PCAP Certified Associate Python Programmer',
        'project': 'Build an Automated Asynchronous ETL Data Pipeline',
        'type': 'Technical Courseware'
    },
    'Docker': {
        'course': 'Enterprise Containerization with Docker & Kubernetes',
        'certification': 'Docker Certified Associate (DCA)',
        'project': 'Containerize Core Microservices and Implement Docker Compose',
        'type': 'DevOps Hands-on'
    },
    'Kubernetes': {
        'course': 'Cloud Native Orchestration & Microservices',
        'certification': 'Certified Kubernetes Administrator (CKA)',
        'project': 'Deploy Zero-Downtime Multi-Cluster Deployment',
        'type': 'Cloud Certification'
    },
    'Sql': {
        'course': 'Advanced SQL Analytics and Query Optimization',
        'certification': 'Oracle / PostgreSQL Certified Professional',
        'project': 'Optimize High-Throughput Relational Database Queries',
        'type': 'Data Engineering'
    },
    'Financial Modeling': {
        'course': 'Advanced Corporate Financial Modeling & Valuation',
        'certification': 'FMVA (Financial Modeling & Valuation Analyst)',
        'project': 'Construct 3-Statement Forecasting and Scenario Model',
        'type': 'Finance Masterclass'
    },
    'Seo': {
        'course': 'Technical SEO & Modern Web Visibility Masterclass',
        'certification': 'Google Analytics & Advanced SEO Certification',
        'project': 'Conduct Technical Site Audit and Core Web Vitals Optimization',
        'type': 'Digital Marketing'
    },
    'Talent Acquisition': {
        'course': 'Strategic Talent Acquisition & Competency-Based Sourcing',
        'certification': 'SHRM-CP / AIRS Certified Diversity Recruiter',
        'project': 'Design High-Retention Executive Hiring Funnel',
        'type': 'HR Leadership'
    },
    'Ci/Cd': {
        'course': 'Continuous Integration & Continuous Deployment with GitHub Actions',
        'certification': 'DevOps Institute CI/CD Foundation',
        'project': 'Architect End-to-End Automated Testing & Deployment Pipeline',
        'type': 'DevOps Hands-on'
    },
    'Default': {
        'course': 'Professional Competency Masterclass',
        'certification': 'Enterprise Skill Certification',
        'project': 'Cross-Functional Capability Capstone Project',
        'type': 'Professional Development'
    }
}

recommendations = []

for idx, row in df_gaps.iterrows():
    emp_id = row['employee_id']
    if pd.notnull(row['missing_skills']) and row['missing_skills'].strip():
        missing_list = [s.strip() for s in row['missing_skills'].split(';') if s.strip()]
        
        for sk in missing_list:
            cat_info = upskilling_catalog.get(sk, upskilling_catalog.get(sk.title(), upskilling_catalog['Default']))
            sev = severity_map.get(sk, "MEDIUM")
            priority = "High" if sev == "HIGH" else ("Medium" if sev == "MEDIUM" else "Low")
            
            recommendations.append({
                "employee_id": emp_id,
                "missing_skill": sk,
                "recommended_course": cat_info['course'],
                "recommended_certification": cat_info['certification'],
                "recommended_project": cat_info['project'],
                "recommendation_type": cat_info['type'],
                "priority": priority
            })

df_recs = pd.DataFrame(recommendations)
df_recs.to_csv(processed_dir / "employee_recommendations.csv", index=False)
print(f"Generated employee_recommendations.csv with {len(df_recs)} targeted upskilling recommendations.")
display(df_recs.head(10))
```
