This notebook maps enterprise organizational roles to canonical O*NET SOC occupations and aggregates required essential and technical software competencies.

```python
import pandas as pd
from pathlib import Path

processed_dir = Path("data/processed")

df_occ = pd.read_csv(processed_dir / "occupation_master.csv")
df_ess = pd.read_csv(processed_dir / "essential_skills_processed.csv")
df_soft = pd.read_csv(processed_dir / "software_skills_processed.csv")

# Standard enterprise organizational roles present in datasets
org_roles = [
    'Account Manager', 'Accountant', 'Auditor', 'Business Development',
    'Content Lead', 'Content Strategist', 'Cybersecurity Specialist',
    'Data Analyst', 'Developer', 'Employee Relations', 'Engineer',
    'Financial Analyst', 'HR Executive', 'HR Manager', 'Helpdesk',
    'Marketing Executive', 'Recruitment Specialist', 'SEO Analyst',
    'SEO Specialist', 'Sales Executive', 'Support Engineer', 'Tester', 'Software Engineer'
]

# Canonical role mapping dictionary to standard role archetypes
role_archetypes = {
    'Developer': ['Python', 'SQL', 'Git', 'Docker', 'REST APIs', 'System Design'],
    'Software Engineer': ['Python', 'Java', 'Data Structures', 'Docker', 'Kubernetes', 'CI/CD', 'Git'],
    'Tester': ['Test Automation', 'Selenium', 'Unit Testing', 'JIRA', 'Python', 'API Testing'],
    'Engineer': ['Cloud Infrastructure', 'Linux', 'Python', 'Networking', 'Security', 'Git'],
    'Support Engineer': ['Troubleshooting', 'Linux', 'SQL', 'Customer Support', 'ITIL', 'Bash Scripting'],
    'Helpdesk': ['Customer Service', 'Hardware Support', 'Windows OS', 'Ticketing Systems', 'Active Directory'],
    'Cybersecurity Specialist': ['Network Security', 'Penetration Testing', 'SIEM', 'Cryptography', 'Python', 'Incident Response'],
    'Data Analyst': ['SQL', 'Python', 'Tableau', 'Power BI', 'Statistical Analysis', 'Excel'],
    'Accountant': ['Financial Accounting', 'Tax Compliance', 'Excel', 'QuickBooks', 'Auditing', 'ERP Systems'],
    'Auditor': ['Internal Auditing', 'Risk Assessment', 'Compliance', 'Financial Analysis', 'Excel', 'Forensic Accounting'],
    'Financial Analyst': ['Financial Modeling', 'Valuation', 'Excel', 'SQL', 'Forecasting', 'Power BI'],
    'HR Manager': ['Talent Acquisition', 'Labor Law', 'Performance Management', 'HRIS', 'Employee Relations', 'Leadership'],
    'HR Executive': ['Recruitment', 'Onboarding', 'HR Operations', 'HRIS', 'Communication'],
    'Employee Relations': ['Conflict Resolution', 'Employment Law', 'Mediation', 'HR Policies', 'Investigation'],
    'Recruitment Specialist': ['Technical Sourcing', 'ATS', 'Interviewing', 'Salary Negotiation', 'Employer Branding'],
    'Sales Executive': ['B2B Sales', 'CRM', 'Negotiation', 'Lead Generation', 'Contract Closing', 'Communication'],
    'Account Manager': ['Client Retention', 'Account Planning', 'CRM (Salesforce)', 'Upselling', 'Contract Negotiation'],
    'Business Development': ['Market Research', 'Partnership Development', 'Pipeline Management', 'Strategic Pitching', 'CRM'],
    'Marketing Executive': ['Digital Marketing', 'Campaign Management', 'Google Analytics', 'Content Marketing', 'SEO'],
    'SEO Analyst': ['Keyword Research', 'Technical SEO', 'Google Search Console', 'Ahrefs / SEMrush', 'Google Analytics'],
    'SEO Specialist': ['On-Page SEO', 'Backlink Strategy', 'Content Optimization', 'Google Analytics', 'HTML / CSS Basics'],
    'Content Lead': ['Content Strategy', 'Editorial Planning', 'Copywriting', 'SEO', 'Brand Storytelling', 'Team Leadership'],
    'Content Strategist': ['Content Auditing', 'Information Architecture', 'Copywriting', 'SEO', 'User Research']
}

# Create canonical role skill master matrix
records = []
for role, skills in role_archetypes.items():
    for sk in skills:
        records.append({
            "job_role": role,
            "required_skill": sk,
            "skill_type": "Technical / Domain",
            "criticality": "Core Required"
        })

df_role_skills = pd.DataFrame(records)
df_role_skills.to_csv(processed_dir / "role_skill_matrix.csv", index=False)
print(f"Created canonical role skill matrix with {len(df_role_skills)} mappings across {len(role_archetypes)} roles.")
display(df_role_skills.head(10))
```
